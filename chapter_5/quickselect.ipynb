```python
import random
import math
# algorytm na zwykla medniane
def median(t:list):
    t.sort()
    t_len = len(t)
    if len(t) % 2 == 1:
        return t[t_len//2]
    else:
        return (t[t_len//2 - 1] + t[t_len//2]) / 2

import math
import random
# algorytm na mediane o lepszej zlozonosci obliczeniowej 
def quick_select_median(l: list):
    num_of_rec = 0
    def quick_select(arr, k):
        pivot = arr[int(random.uniform(0, len(arr)))]
        left_p = []
        right_p = []
        middle = []
        
        nonlocal num_of_rec
        num_of_rec += 1

        for element in arr:
            if element < pivot:
                left_p.append(element)
            elif element == pivot:    
                middle.append(element)
            else:
                right_p.append(element)
        
        if k < len(left_p):
            return quick_select(left_p, k)
        elif k < len(left_p) + len(middle):
            return pivot
        else:
            nowe_k = k - len(left_p) - len(middle)
            return quick_select(right_p, nowe_k)
            
    if len(l) & 2 == 1:
        target_k = math.floor(len(l) / 2)
        return quick_select(l, target_k)
    else: 
        left_k = (len(l) // 2) - 1
        right_k = len(l) // 2
        return (quick_select(l , left_k) + quick_select(l, right_k)) / 2, num_of_rec 

print(quick_select_median([1, 1, 3, 4, 5, 1]))
print(median([1, 1, 3, 4, 5, 1]))

```

```python
list_of_n = [i for i in range(50_000, 1_000_001, 50_000)]

def random_uniform_list(n):
    return [random.uniform(0, 1_000_000) for _ in range(n)]


liczba_symulacji = 50 
num_of_rec = []

for n in list_of_n:
    suma_rekurencji = 0
    
    for _ in range(liczba_symulacji):
        suma_rekurencji += quick_select_median(random_uniform_list(n))[1]
    
    srednia_rekurencji = suma_rekurencji / liczba_symulacji
    num_of_rec.append(srednia_rekurencji)
    
    print(f"Obliczono dla n = {n} (Średnia: {srednia_rekurencji})")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.scatter(list_of_n, num_of_rec)
plt.grid(True)
plt.show()
```
