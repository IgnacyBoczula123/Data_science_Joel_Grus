```python
# klasa opisujaca kliker

class CountingClicer:
    
    def __init__(self, count = 0):
        self.count = count
    
    def __repr__(self):
        return f'clicker(count={self.count})'

    def click(self):
        self.count += 1
    
    def reset(self):
        self.count = 0
    
    def read(self):
        return self.count
```

```python
# testy przy uzyciu assert
clicker = CountingClicer()
assert clicker.read() == 0, 'should be zero'
clicker.click()
clicker.click()
assert clicker.read() == 2, 'should be two'
clicker.reset()
assert clicker.read() == 0, 'should be zero'
```
