```python
X = TypeVar('X')
Y = TypeVar('Y')
def split_data(data: List[X], prob: float) -> Tuple[List[X], List[X]]:
    """Split data into fractions [prob, 1 - prob]"""
    data = data[:]                    # Make a shallow copy
    random.shuffle(data)              # because shuffle modifies the list.
    cut = int(len(data) * prob)       # Use prob to find a cutoff
    return data[:cut], data[cut:]     # and split the shuffled list there.

def train_test_split(xs: List[X],
                     ys: List[Y],
                     test_pct: float) -> Tuple[List[X], List[X], List[Y], List[Y]]:
    # Generate the indices and split them.
    idxs = [i for i in range(len(xs))]
    train_idxs, test_idxs = split_data(idxs, 1 - test_pct)

    return ([xs[i] for i in train_idxs],  # x_train
            [xs[i] for i in test_idxs],   # x_test
            [ys[i] for i in train_idxs],  # y_train
            [ys[i] for i in test_idxs])   # y_test

def scale(data):
    data = np.array(data)
    return np.mean(data, axis=0), np.std(data, axis=0)

def rescale(data):
    data = np.array(data)
    mean, std = scale(data)
    rescaled = np.zeros_like(data, dtype=float)
    
    for i in range(data.shape[1]):
        if std[i] > 0:
            # Standaryzacja dla zwykłych zmiennych objaśniających
            rescaled[:, i] = (data[:, i] - mean[i]) / std[i]
        else:
            # Pozostawienie kolumny bez zmian (np. jedynki dla wyrazu wolnego)
            rescaled[:, i] = data[:, i]
            
    # Zwracamy listę list, aby reszta Twojego kodu działała bez modyfikacji
    return rescaled.tolist()
```

```python
rescaled_xs = rescale(xs)
random.seed(0)
x_train, x_test, y_train, y_test = train_test_split(rescaled_xs, ys, 0.33)

learning_rate = 0.01

# pick a random starting point
beta = [random.random() for _ in range(3)]

with tqdm.trange(5000) as t:
    for epoch in t:
        gradient = negative_log_gradient(x_train, y_train, beta)
        beta = [b - learning_rate * g for b, g in zip(beta, gradient)]
        loss = negative_log_likelihood(x_train, y_train, beta)
        t.set_description(f"loss: {loss:.3f} beta: {beta}")
```

```python
true_positives = false_positives = true_negatives = false_negatives = 0

for x_i, y_i in zip(x_test, y_test):
    prediction = logistic(np.dot(beta, x_i))
    if y_i == 1 and prediction >= 0.5:  # TP: paid and we predict paid
        true_positives += 1
    elif y_i == 1:                      # FN: paid and we predict unpaid
        false_negatives += 1
    elif prediction >= 0.5:             # FP: unpaid and we predict paid
        false_positives += 1
    else:                               # TN: unpaid and we predict unpaid
        true_negatives += 1

precision = true_positives / (true_positives + false_positives)
recall = true_positives / (true_positives + false_negatives)



print(precision, recall)
```
