```python
def add(v, w):
    assert len(v) == len(w), 'vectors must be same size'
    return[v_i + w_i for v_i, w_i in zip(v,w)]
assert add([1,1,1], [2,2,2]) == [3,3,3]
```

```python
# pomysl na stworzenie macierzy przujaciol 

friendship_pairs = [
    (0, 1), (0, 2), (1, 2), (1, 3), (2, 3), (3, 4),
    (4, 5), (5, 6), (5, 7), (6, 8), (7, 8), (8, 9)
]

friendship_dict = {i: [] for i in range(10)}

for i in range(len(friendship_pairs)):
    
    friend_left = friendship_pairs[i][0]
    friend_right = friendship_pairs[i][1]
    
    friendship_dict[friend_left].append(friend_right)
    friendship_dict[friend_right].append(friend_left)

def make_matrix(num_rows, num_colums, entry_func):
    return [
            [entry_func(i, j) for j in range(num_colums)] 
            for i in range(num_rows)]

friend_matrix = make_matrix(len(friendship_dict.keys()), len(friendship_dict.keys()), lambda i,j: 1 if i in friendship_dict[j] else 0)
# licza przyjacol osoby id 2
sum(friend_matrix[2])
friend_matrix 
'''Wlasnosci: 
    symetria
    sprawdzanie przyjazni w O(1)
    A^2 = A*A - wartosc [i,j] to liczba wspolnych znajomych i oraz j
    A^3 znajomi znajomych '''
```

```python
# wykorzystnie wlasnosci
import numpy as np

friend_array = np.array(friend_matrix)
mutual_friends_array = np.linalg.matrix_power(friend_array, 2)
mutual_friends_array

np.linalg.matrix_power(friend_array, 3)
```
