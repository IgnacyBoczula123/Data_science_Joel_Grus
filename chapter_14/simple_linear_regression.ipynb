```python
# prosty model do przewidywania czasu przedzonego na stronie na podstawie ilosci znajomych 
from typing import List, Tuple
import numpy as np
def predict(alpha: float, beta:float, x_i: float) -> float:
    
    y_i = beta * x_i + alpha
    return y_i

def error(alpha: float, beta:float, x_i: float, y_i) -> float:
    
    y_pred = predict(alpha, beta, x_i)
    error = y_pred - y_i
    return error

def sum_of_sqrerrors(alpha: float, beta: float, x: List[float], y:List[float]) -> float:
    return sum([(error(alpha, beta, x_i, y_i))**2 for y_i, x_i in zip(y,x)])

def least_squares_fit(x: List[float], y: List[float]) -> Tuple[float, float]:
    'na podstawie x i y znajdz alpha i beta gdzie sqrerr jest najmniejszy'
    
    r = np.corrcoef(x, y)[0, 1]
    
    beta = float(r * np.std(y) / np.std(x))
    alpha = float(np.mean(y) - beta * np.mean(x))
    
    return alpha, beta

# test
x = list(np.linspace(-100, 100, 10))
y = [3 * float(i) - 5 for i in x]
print(least_squares_fit(x, y))
```

```python
def total_sum_of_squares(y: List) -> float:
    """the total squared variation of y_i's from their mean"""
    mean_y = np.mean(y)
    de_mean = [i - mean_y for i in y]
    return sum(v ** 2 for v in de_mean)

def r_squared(alpha: float, beta: float, x, y) -> float:
    """
    the fraction of variation in y captured by the model, which equals
    1 - the fraction of variation in y not captured by the model
    """
    return 1.0 - (sum_of_sqrerrors(alpha, beta, x, y) /
                  total_sum_of_squares(y))
print(r_squared(-5, 3, x, y))
y_noise = [3 * float(i) - 5 + np.random.normal(0, 250) for i in x]
x_n, y_n = least_squares_fit(x, y_noise)
print(r_squared(x_n, y_n, x, y_noise))
```
