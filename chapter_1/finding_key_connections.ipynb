```python
users = [
    { "id": 0, "name": "Hero" },
    { "id": 1, "name": "Dunn" },
    { "id": 2, "name": "Sue" },
    { "id": 3, "name": "Chi" },
    { "id": 4, "name": "Thor" },
    { "id": 5, "name": "Clive" },
    { "id": 6, "name": "Hicks" },
    { "id": 7, "name": "Devin" },
    { "id": 8, "name": "Kate" },
    { "id": 9, "name": "Klein" },
]

friendship_pairs = [
    (0, 1), (0, 2), (1, 2), (1, 3), (2, 3), (3, 4),
    (4, 5), (5, 6), (5, 7), (6, 8), (7, 8), (8, 9)
]
```

```python
# chcemy stworzyc dist gdzie klucz to id a wartosc to lista przyjaciol tego id 
# implementacja wlasana 
friendship_dict = {user['id']: [] for user in users}

for i in range(len(friendship_pairs)):
    
    friend_left = friendship_pairs[i][0]
    friend_right = friendship_pairs[i][1]
    
    friendship_dict[friend_left].append(friend_right)
    friendship_dict[friend_right].append(friend_left)

friendship_dict
```

```python
# chcemy znalesc sume wszystkich polaczej oraz liczbe friends dla konkretnego usera 
def num_of_friends(user: int):
    return len(friendship_dict[user])

total_connections = sum(num_of_friends(user['id']) for user in users)
print(f'total connections: {total_connections}\navg_conn {total_connections/ len(friendship_dict.keys())}')
```

```python
# lista id od najwiekszej ilosci polaczej do najmniejszej 

l = [(user['id'], num_of_friends(user['id'])) for user in users]
sorted_list = sorted(l, key=lambda x: x[1], reverse=True) # do key tafia kazdy tuple a lambda bierze z drugi element 
sorted_list
```

```python
from collections import Counter
def fof(user):
    first_friends = friendship_dict[user]    
    second_friends = []
    
    for f_f in first_friends:
        second_friends += friendship_dict[f_f]
    
    for s_f in second_friends.copy():
        if s_f in first_friends or s_f == user:
            second_friends.remove(s_f)
    
    return Counter(second_friends)
fof(3)
```

```python
# znalesc osobe o wybranym zainteresowaniu
def find_id_intrest(interest :str):
    return [id 
            for id, user_interest in interests 
            if user_interest == interest]
find_id_intrest('Java')
```

```python
from collections import defaultdict

user_ids_by_interest = defaultdict(list)

for user_id, interest in interests:
    user_ids_by_interest[interest].append(user_id) # dzieki defaultdict nie ma bledu o braku klucza 

interests_by_user_id = defaultdict(list)

for user_id, interest in interests:
    interests_by_user_id[user_id].append(interest)
```

```python
def most_common_interests_with(user):
    return Counter(
        interested_user_id
        for interest in interests_by_user_id[user["id"]]
        for interested_user_id in user_ids_by_interest[interest]
        if interested_user_id != user["id"]
    )
most_common_interests_with(users[0])
```
